#Exercício 1 – Condicional Simples

Crie um programa que pergunte ao usuário se ele já montou sua árvore de Natal.

Se a resposta for “sim”, exiba uma mensagem motivacional.
Se a resposta for “não”, exiba uma mensagem encorajando-o a montar a árvore hoje.


```python
def arvore():
    resposta = input("Você já montou sua árvore de Natal? \nDigite 1 para sim \nDigite 2 para não\n")

    if resposta == "1":
        print("Que ótimo! Aproveite esse clima especial e continue espalhando alegria! 🎄✨")
    elif resposta == "2":
        print("Então hoje é um ótimo dia para montar sua árvore e entrar no clima natalino! 🎅🎄")
    else:
        print("Resposta não reconhecida. Tente novamente.")
```

```python
arvore()
```

# Exercício 2 – Laço de Repetição

Peça ao usuário um número N e imprima uma contagem regressiva de N até 1, simulando um cronômetro para a chegada do Natal.



```python
n = int(input("Digite um número para que seja impressa a contagem regressiva, até 1"))
print("Faltam", n, "dias para o Natal")

while (n > 1):
  n = n - 1

  if n == 1:
    print("Falta",n,"dia para o Natal!")

  else:
    print("Faltam",n,"dias para o Natal!")


```

# Exercício 3 – Funções + Lista + Condicional

Crie uma função chamada verificar_presentes(lista_presentes) que recebe uma lista de presentes desejados para o Natal.
A função deve:

Verificar se a lista tem pelo menos 3 itens.
Exibir uma mensagem adequada:
Se tiver 3 ou mais itens → “O Papai Noel terá trabalho este ano!”
Caso contrário → “Sua lista está muito curta!”
Retornar a quantidade total de presentes.


```python
lista_presentes = ["boneca", "carrinho",]

def verificar_presentes(lista_presentes):

    if len(lista_presentes) >= 3:
      print("O Papai Noel terá trabalho este ano!")

    else:
      print("Sua lista está muito curta!")


    return len(lista_presentes)
```

```python
verificar_presentes(lista_presentes)
```

# Exercício 4 – Funções + Dicionário + Laços

Crie uma função contar_enfeites(enfeites) que recebe um dicionário onde:

chave → tipo de enfeite de Natal
valor → quantidade daquele enfeite
A função deve percorrer o dicionário usando um laço de repetição, somar todas as quantidades e retornar o total de enfeites.

```python
enfeites = {"Pisca-Pica": 25, "Arvore": 350, "Presentes Falsos": 60}

def contar_enfeites(enfeites):
  total = 0
  for i in enfeites.values():
    total = total + i

  print("O total gasto em enfeites é:", total)
```

```python
contar_enfeites(enfeites)
```

#  Exercício 5 – Função Completa (listas, dicionários, laço e condicional)

Crie um sistema simples para organizar uma ceia de Natal:

Peça ao usuário os pratos que cada pessoa da família vai levar (use laço para cadastrar).
Armazene os dados em um dicionário, onde:
chave → nome da pessoa
valor → lista de pratos que ela levará
Crie a função resumo_ceia(dados) que deve mostrar:
Quantas pessoas irão contribuir
Quantos pratos no total serão levados
Se houver mais de 10 pratos, exiba: “Essa ceia vai ser farta!”

```python
dados = {}

while True:
    nome = input("Nome da pessoa (ou 'fim' para encerrar): ")

    if nome.lower() == "fim":
        break

    pratos = []
    print(f"Cadastre os pratos que {nome} vai levar:")

    while True:
        prato = input("  Digite um prato (ou 'ok' para terminar): ")
        if prato.lower() == "ok":
            break
        pratos.append(prato)

    dados[nome] = pratos



def resumo_ceia(dados):
    total_pessoas = len(dados)

    total_pratos = 0
    for pratos in dados.values():
        total_pratos += len(pratos)

    print("\n----- RESUMO DA CEIA -----")
    print("Pessoas contribuindo:", total_pessoas)
    print("Total de pratos:", total_pratos)

    if total_pratos > 10:
        print("Essa ceia vai ser farta!")



```

```python
resumo_ceia(dados)
```
